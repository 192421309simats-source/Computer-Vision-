```python
from google.colab import files

uploaded = files.upload()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]

image = cv2.imread(filename)

print("Image uploaded successfully:", filename)
print("Original Image Size:", image.shape)
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
bigger = cv2.resize(image, None, fx=2, fy=2,
                    interpolation=cv2.INTER_LINEAR)

print("Bigger Image Size:", bigger.shape)
```

```python
plt.imshow(cv2.cvtColor(bigger, cv2.COLOR_BGR2RGB))
plt.title("Bigger Image")
plt.axis("off")
plt.show()
```

```python
smaller = cv2.resize(image, None, fx=0.5, fy=0.5,
                     interpolation=cv2.INTER_AREA)

print("Smaller Image Size:", smaller.shape)
```

```python
plt.imshow(cv2.cvtColor(smaller, cv2.COLOR_BGR2RGB))
plt.title("Smaller Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(bigger, cv2.COLOR_BGR2RGB))
plt.title("Bigger Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(smaller, cv2.COLOR_BGR2RGB))
plt.title("Smaller Image")
plt.axis("off")

plt.show()
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]

image = cv2.imread(filename)

print("Image uploaded successfully:", filename)
```

```python
def analyze_histogram(image):
    colors = ('b', 'g', 'r')
    channel_names = ('Blue', 'Green', 'Red')

    for i in range(3):
        histogram = cv2.calcHist([image], [i], None, [256], [0, 256])

        print(channel_names[i], "Channel")
        print("Minimum:", int(histogram.min()))
        print("Maximum:", int(histogram.max()))
        print("Total Pixels:", int(histogram.sum()))
        print()

        plt.plot(histogram, color=colors[i])

    plt.title("Color Histogram")
    plt.xlabel("Color Level (0-255)")
    plt.ylabel("Number of Pixels")
    plt.show()
```

```python
analyze_histogram(image)
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
import cv2
from IPython.display import Video, display
```

```python
filename = list(uploaded.keys())[0]

cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print("Video uploaded successfully:", filename)
print("FPS:", fps)
print("Width:", width)
print("Height:", height)

cap.release()
```

```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

slow_video = cv2.VideoWriter(
    "slow_motion.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps / 2,
    (width, height)
)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    slow_video.write(frame)

cap.release()
slow_video.release()

print("Slow motion video created successfully.")
```

```python
display(Video("slow_motion.mp4", embed=True))
```

```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fast_video = cv2.VideoWriter(
    "fast_motion.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps * 2,
    (width, height)
)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    fast_video.write(frame)

cap.release()
fast_video.release()

print("Fast motion video created successfully.")
```

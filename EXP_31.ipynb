```python

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload image dynamically
uploaded = files.upload()

# Step 2: Read input image
image_path = list(uploaded.keys())[0]
img = cv2.imread(image_path)

if img is None:
    raise ValueError("Unable to load image!")

# Step 3: Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Step 4: Get threshold values dynamically
lower = int(input("Enter lower threshold value (0-255): "))
upper = int(input("Enter upper threshold value (0-255): "))

if not (0 <= lower <= 255 and 0 <= upper <= 255):
    raise ValueError("Threshold values must be between 0 and 255!")

if lower > upper:
    raise ValueError("Lower threshold must be less than or equal to upper threshold!")

# Step 5: Apply threshold-based segmentation
segmented = cv2.inRange(gray, lower, upper)

# Step 6: Apply mask to original image
result = cv2.bitwise_and(img, img, mask=segmented)

# Step 7: Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(segmented, cmap="gray")
plt.title("Segmentation Mask")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Segmented Image")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Image segmentation completed successfully!")
print("Lower Threshold:", lower)
print("Upper Threshold:", upper)
```
